```python
#function with loop
def get_random_point_t(radius,num):
    arr2=np.array([])
    for i in range(num):
        r = random.uniform(0,400)
        theta = random.uniform(0,2*math.pi)
        x1 = r*math.cos(theta)
        y1 = r*math.sin(theta)
        r2 = random.uniform(0,400)
        theta2 = random.uniform(0,2*math.pi)
        x2 = r2*math.cos(theta2)
        y2 = r2*math.sin(theta2)

        a = x1-x2
        b = y1-y2
        c = math.sqrt((a * a) + (b * b))
        arr2=np.append(arr2,c)
    return (arr2)


```

```python
def get_random_point_r(radius):
    
    r = random.uniform(0,radius)
    theta = random.uniform(0,2*math.pi)
    x1 = r*math.cos(theta)
    y1 = r*math.sin(theta)
    r2 = random.uniform(0,radius)
    theta2 = random.uniform(0,2*math.pi)
    x2 = r2*math.cos(theta2)
    y2 = r2*math.sin(theta2)

    a = x1-x2
    b = y1-y2
    c = math.sqrt((a * a) + (b * b))
    return (c)
```

```python
#generate two dot
def get_random_point(radius):
    
    while True:
        # Generate the random point
        
        x = random.uniform(-radius+5, radius-6)
        y = random.uniform(-radius+5, radius-6)
        # Check that it is inside the circle
        if math.sqrt(x ** 2 + y ** 2) < radius:
            # Return it
        return(x,y)
    
```

```python

arr2 = np.array([])
random.seed(100000)
for i in range(100000):
    x1,y1=get_random_point(400)
    x2,y2=get_random_point(400)
    a = x1-x2
    b = y1-y2
    c = math.sqrt((a * a) + (b * b))
    arr2=np.append(arr2,c)


plt.hist(arr2, bins=400,
         alpha=0.5, density=1,
         label='Random Data')

plt.xlabel('distance')
plt.ylabel('frequency')
plt.legend()
plt.title('Distance Histogram: DGRP total')
plt.show
```
